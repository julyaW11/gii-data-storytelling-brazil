```python
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
import numpy as np
from IPython.display import HTML

anos = np.array([2011, 2012, 2013, 2014, 2015, 2016, 2017, 2018, 2019, 2020, 2021, 2022, 2023, 2024])
posicoes = np.array([47, 59, 64, 61, 70, 69, 69, 64, 66, 62, 57, 54, 49, 50])

fig, ax = plt.subplots(figsize=(10, 6))


ax.set_ylim(posicoes.max() + 5, posicoes.min() - 5)
ax.set_xlim(anos.min() - 1, anos.max() + 1)


ax.invert_yaxis()


ax.set_title('Posições do Brasil no GII desde 2011', fontsize=16)
ax.set_ylabel('Posição no Ranking')
ax.set_xlabel('Ano')


ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)


ax.set_xticks(anos)
plt.xticks(rotation=45)


line, = ax.plot([], [], lw=2, color='#c43a3a', marker='o', markersize=5)
scatter = ax.scatter([], [], s=50, color='#c43a3a')
text_labels = [ax.text(anos[i], posicoes[i] - 1.2, f'{posicoes[i]}º',
                       ha='center', va='bottom', fontsize=9, fontweight='bold', visible=False) for i in range(len(anos))]


def init():
    line.set_data([], [])

    scatter.set_offsets(np.c_[[], []])
    for label in text_labels:
        label.set_visible(False)
    return line, scatter, *text_labels

# Função de Animação ---
def animate(frame):
    x_data = anos[:frame+1]
    y_data = posicoes[:frame+1]

    line.set_data(x_data, y_data)
    scatter.set_offsets(np.c_[x_data, y_data])


    for i in range(frame + 1):
        text_labels[i].set_visible(True)

    return line, scatter, *text_labels


ani = FuncAnimation(fig, animate, frames=len(anos), init_func=init,
                    interval=800, blit=True, repeat=False)


plt.close()
HTML(ani.to_html5_video())
```
